The data is numerical. The groups(male and female) are independent of each others height. There are two groups.

Test for normality Ks (Kolmogorov-Smirnov) normality test beacause data points are about 50


```python
import pandas as pd
import matplotlib.pyplot as plt
athletes = pd.read_csv("https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2021/2021-07-27/athlete_events.csv")
df = athletes[['Height', "Sex"]].dropna()
df.head()
```

In the KS test we make the null hypothesis that the data is normally distributed and we compare it to the theoretical normal distribution.

```python

height_female = df[df["Sex"] == "F"]["Height"]
height_male = df[df["Sex"] == "M"]["Height"]
height_male.head()

```

```python
from scipy import stats
normal_ditribution_analysis = stats.ks_2samp(height_female, height_male)
print(normal_ditribution_analysis)
```

We therefore reject that null hypothesis that the data was obtained from a normal dataset.

Equality of variance

```python
equality_of_variance = stats.bartlett(height_female, height_male)
equality_of_variance
```

State the justifications for your choice for a statistical test.

1. The Kolmogorov-Smirnov (KS) is normality test that calculates to see whether the data has been obtained from a normally ditributed data. It can check for a one sample of data in a population. It can also to check for two sample of data. it produces a p-value which should be validated with signficance level(alpha).The null hypothesis that holds at in the KS test is that the data is from a normally distributed dataset. If the value is less than the significance valBartletta thee we reject the null hypothesis
2. The bartlett is look for the equivalence of variance. This check whether the sample being tensted has the same variance all across the population


Conduct the statistical test.

```python
mean_statistical_test = stats.ttest_ind(height_female, height_male, equal_var=True)
stats = mean_statistical_test.statistic
p_value = mean_statistical_test.pvalue
print(stats)
print(p_value)
```

Since the p-value is less than the significance value this means  that we reject the null hypothesis; height of the male athletes and female athletes  is not equal. We therefore adopt the alternative hypothesis.
